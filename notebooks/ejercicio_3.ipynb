```python
import pandas as pd
import numpy as np
from src.lista4.data import cargar_datos
from src.lista4.models import ajustar_regresion_lineal
from src.lista4.metrics import riesgo_empirico_cuadratico
```

Carga y transformacion

```python
df = cargar_datos('../data/experimento_3.csv')
df['z2'] = df['x2'] / 100
y = df['y']
```

Ajuste modelo 1 y 2

```python
mod_beta = ajustar_regresion_lineal(df[['x1', 'x2']], y)
mod_gamma = ajustar_regresion_lineal(df[['x1', 'z2']], y)

pred_beta = mod_beta.predict(df[['x1', 'x2']])
pred_gamma = mod_gamma.predict(df[['x1', 'z2']])

riesgo_beta = riesgo_empirico_cuadratico(y, pred_beta)
riesgo_gamma = riesgo_empirico_cuadratico(y, pred_gamma)
```

Max $d_i$

```python
d_i = np.abs(pred_beta - pred_gamma)
max_di = np.max(d_i)
```

Verificacion numerica (errores absolutos)

```python
err_b0 = np.abs(mod_gamma.intercept_ - mod_beta.intercept_)
err_b1 = np.abs(mod_gamma.coef_[0] - mod_beta.coef_[0])
err_b2 = np.abs(mod_gamma.coef_[1] - (100 * mod_beta.coef_[1]))
print(f"Error absoluto intercepciones: {err_b0}")
print(f"Error absoluto betas_1: {err_b1}")
print(f"Error absoluto betas_2: {err_b2}")
```

Tabla resumen

```python
tabla_ej3 = pd.DataFrame({
    'Modelo': ['Original', 'Escalado (z2)'],
    'Coeficientes': [list(mod_beta.coef_), list(mod_gamma.coef_)],
    'Riesgo Empírico': [riesgo_beta, riesgo_gamma],
    'Max d_i': [None, max_di]
})
display(tabla_ej3)
```
